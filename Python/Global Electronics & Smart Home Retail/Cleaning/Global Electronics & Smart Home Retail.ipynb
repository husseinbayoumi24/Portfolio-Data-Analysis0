Cleaning Table Products

```python
products.head(5)
```

```python
products.info()
```

```python
c_products=products.copy()
```

```python
c_products["UnitPrice"].isna().sum()
```

```python
invalid_prices = products[products['UnitPrice'] <= 0]
len(invalid_prices)
```

```python
c_products.loc[c_products['UnitPrice'] <= 0, 'UnitPrice'] = np.nan
```

```python
c_products["UnitPrice"]=(
    c_products["UnitPrice"].fillna(
        c_products.groupby("Category")["UnitPrice"].transform("mean")
    ))
```

```python
c_products.info()
```

```python
mask=(c_products["StandardCost"] > c_products["UnitPrice"]).fillna(False)
len(mask)
```

يمكننا جعل سعر البيع يساوي تكلفة التصنيع مضافاً إليها هامش ربح بسيط (مثلاً 20% كحد أدنى):

```python
c_products.loc[mask, 'UnitPrice'] = c_products.loc[mask, 'StandardCost'] * 1.20
```

```python
c_products["Category"].unique()
```

```python
# إزالة المسافات الزائدة من الأطراف لجميع الأعمدة النصية في جدول المنتجات
text_columns = c_products.select_dtypes(include=['object']).columns
for col in text_columns:
    c_products[col] = c_products[col].astype(str).str.strip()

print("تم تنظيف وإزالة المسافات الزائدة بنجاح!")
```

Cleaning Table Orders

```python
c_orders=orders.copy()
```

```python
c_orders.head(5)
```

```python
c_orders.info()
```

```python
c_orders["OrderDate"]=pd.to_datetime(
    c_orders["OrderDate"],
    errors="coerce")

c_orders["ShippingDate"]=pd.to_datetime(
    c_orders["ShippingDate"],
    errors="coerce"
)
```

```python
invalid_dates=(c_orders[c_orders["ShippingDate"] < c_orders["OrderDate"]])
len(invalid_dates)
```

```python
c_orders= c_orders[~(c_orders["ShippingDate"] < c_orders["OrderDate"])]
```

```python
c_orders["SalesChannel"]=c_orders["SalesChannel"].str.strip()
c_orders["SalesChannel"].unique()
```

```python
channel_map={
    'online':'Online',
    'Web':'Online',
    'store':'Store'
}
c_orders["SalesChannel"]=c_orders["SalesChannel"].replace(channel_map)
```

```python
c_orders.isnull().sum()
```

```python
# حذف الصفوف التي تحتوي على قيم مفقودة في الأرقام الأساسية
c_orders.dropna(subset=['OrderID', 'CustomerID'], inplace=True)
```
